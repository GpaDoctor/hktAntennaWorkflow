````python
import requests

proxies = {
    'http': 'http://proxy.pccw.com:8080',
    'https': 'http://proxy.pccw.com:8080'
}

headers = {
    'x-api-key': x_api_key
}

def bot_builder_chat(data: dict, proxy=False):

    url = f"https://api.{env}.bot-builder.pccw.com/v1/groups/{group_id}/llm-model/chat"
    
    try:
        with requests.post(
            url, 
            json=data,
            headers=headers,
            proxies=proxies if proxy else None, 
            verify=True,
        ) as response:
        
            if response.status_code != 200:
                raise Exception(response.text)
        
            return response.json()
    except Exception as e:
        raise e


def upload_file(file_path, proxy=False):

    url = f"https://api.{env}.bot-builder.pccw.com/v1/groups/{group_id}/chat-files?chatApproach=cwf"
    
    with open(file_path, 'rb') as file:
        
        try:
            files = {'file': file}

            with requests.post(
                url, 
                files=files, 
                headers=headers,
                proxies=proxies if proxy else None, 
                verify=True,
            ) as response:
            
                return response.json()
        except Exception as e:
            raise e
        
        
def bot_builder_embeddings(data: dict, proxy=False):

    url = f"https://api.{env}.bot-builder.pccw.com/v1/groups/{group_id}/llm-model/embeddings"
    
    try:
        with requests.post(
            url, 
            json=data,
            headers=headers,
            proxies=proxies if proxy else None, 
            verify=True,
        ) as response:
        
            if response.status_code != 200:
                raise Exception(response.text)
        
            return response.json()
    except Exception as e:
        raise e
    
def bot_builder_speech(    
    data: dict,
    convert='stt',
    proxy=False,
):

    url = f"https://api.{env}.bot-builder.pccw.com/v1/groups/{group_id}/llm-model/azure-speech/{convert}"
    
    try:
        with requests.post(
            url, 
            json=data,
            headers=headers,
            proxies=proxies if proxy else None, 
            verify=True,
        ) as response:
        
            if response.status_code != 200:
                raise Exception(response.text)
        
            return response.json()
    except Exception as e:
        raise e
````

## Part 3: Chat with file

````python
# edit your file path for file upload
file_path = "test.pdf"

response = upload_file(file_path, proxy)

print(response)

# pass the s3Basename to chat with file to start chat
s3Basename = response.get('s3Basename')

data = {
    "approach": "cwf",
    "overrides": {
        "top": 0,
        "model": "vertexai-gemini-1.5-flash-001",
        "prompt_template": "",
        "max_tokens": 1000,
        "temperature": 0,
        "top_p": 0.95,
        "presence_penalty": 0,
        "frequency_penalty": 0
    },
    "files" : [
        {
            "data_source": "upload",
            "file_id": s3Basename
        }
    ]
}

chat_with_file_queries = [
    'what do you find in the file'
]

for query in chat_with_file_queries:
    
    messages = [
        {
            "role": "user",
            "content": query,
        }
    ]
    data['history'] = messages
    
    try:

        rst = bot_builder_chat(data=data, proxy=proxy)
        
        print(f"Chat with file answer: {rst.get('answer')}")
    except Exception as e:
        print(e)
        
````
